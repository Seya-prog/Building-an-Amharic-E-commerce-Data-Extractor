```python
# =====================================================
# YOLOv8 Object Detection using Public COCO128 Dataset
# =====================================================

# Install
!pip install -q ultralytics

from ultralytics import YOLO
from ultralytics.utils.downloads import download
from PIL import Image
import matplotlib.pyplot as plt
import glob
import os

# -----------------------------------------------------
# Download Dataset
# -----------------------------------------------------

if not os.path.exists("coco128"):
    download("https://github.com/ultralytics/assets/releases/download/v0.0.0/coco128.zip")
    !unzip -q coco128.zip

print("Dataset downloaded successfully!")

# -----------------------------------------------------
# Load Pretrained YOLOv8 Nano
# -----------------------------------------------------

model = YOLO("yolov8n.pt")

# -----------------------------------------------------
# Train Model
# -----------------------------------------------------

model.train(
    data="coco128.yaml",
    epochs=30,
    imgsz=640,
    batch=16,
    lr0=0.001,
    project="YOLO_Assignment",
    name="Experiment",
    pretrained=True
)

# -----------------------------------------------------
# Validate Model
# -----------------------------------------------------

metrics = model.val()

print("\n==============================")
print("Evaluation Metrics")
print("==============================")
print(f"Precision : {metrics.box.mp:.4f}")
print(f"Recall    : {metrics.box.mr:.4f}")
print(f"mAP@0.5   : {metrics.box.map50:.4f}")
print(f"mAP@0.5:0.95 : {metrics.box.map:.4f}")

# -----------------------------------------------------
# Show Training Curves
# -----------------------------------------------------

results_folder = "YOLO_Assignment/Experiment"

files = [
    "results.png",
    "confusion_matrix.png",
    "PR_curve.png",
    "F1_curve.png",
    "P_curve.png",
    "R_curve.png"
]

for f in files:
    path = os.path.join(results_folder, f)
    if os.path.exists(path):
        print(f"\nDisplaying {f}")
        plt.figure(figsize=(10,8))
        plt.imshow(Image.open(path))
        plt.axis("off")
        plt.show()

# -----------------------------------------------------
# Predict on Test Images
# -----------------------------------------------------

print("\nRunning Detection...")

model.predict(
    source="coco128/images/train2017",
    conf=0.25,
    save=True
)

# -----------------------------------------------------
# Display Some Predictions
# -----------------------------------------------------

prediction_images = glob.glob("runs/detect/predict/*.jpg")

print(f"Found {len(prediction_images)} prediction images.")

for img in prediction_images[:5]:
    plt.figure(figsize=(10,8))
    plt.imshow(Image.open(img))
    plt.axis("off")
    plt.show()

# -----------------------------------------------------
# Model Information
# -----------------------------------------------------

print("\n==============================")
print("Model Summary")
print("==============================")

model.info()

print("\n==============================")
print("Hyperparameters Used")
print("==============================")
print("Model          : YOLOv8n")
print("Dataset        : COCO128")
print("Classes Used   : Person, Car, Bottle (among COCO classes)")
print("Epochs         : 30")
print("Batch Size     : 16")
print("Image Size     : 640")
print("Learning Rate  : 0.001")
print("Confidence     : 0.25")

print("\nTraining Completed Successfully!")
```
